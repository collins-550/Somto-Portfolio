```python
class Restaurant:
    def __init__(self, name):
        self.name = name
        self.menu = {
            "Jollof Rice": 1500,
            "Fried Rice": 1800,
            "Chicken": 1000,
            "Pepsi": 500,
            "Water": 200
        }
        self.order = []
        self.total = 0

    def show_menu(self):
        print(f"\n===== {self.name} MENU =====")
        for item, price in self.menu.items():
            print(f"{item}: ₦{price}")
        print("=" * 25)

    def add_order(self, item):
        if item in self.menu:
            self.order.append(item)
            self.total += self.menu[item]
            print(f"{item} added! ₦{self.menu[item]}")
        else:
            print(f"{item} not on menu!")

    def show_bill(self):
        print("\n===== YOUR ORDER =====")
        for item in self.order:
            print(f"- {item}: ₦{self.menu[item]}")
        print(f"TOTAL: ₦{self.total}")
        print("=" * 25)



resto = Restaurant("Somto's Kitchen")

while True:
    resto.show_menu()
    choice = input("\nEnter item to order (or 'done' to finish): ")
    if choice == "done":
        break
    resto.add_order(choice)

resto.show_bill()
```
